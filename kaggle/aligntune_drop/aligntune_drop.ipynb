# AlignTune DPO

The Track 1 chain is CuratorKIT (data) to AlignTune (post-training) to AuditKIT (evaluation).
This kernel trains one DPO adapter through AlignTune's `create_rl_trainer`, after a 5-step smoke.

If the smoke fails, the kernel stops and the TRL path in `kaggle/train` remains the fallback.

```python
import json, os, shutil, subprocess, sys
from pathlib import Path

def find_repo():
    hits = [p for p in Path("/kaggle/input").rglob("rule_checker.py")
            if p.exists() and (p.parents[2] / "scripts" / "train.py").exists()]
    exact = [p for p in hits if "datasets/rockerleo/htd-code" in str(p)]
    hits = exact or hits
    hits.sort(key=lambda p: len(p.parts))
    if not hits:
        raise SystemExit("code dataset not mounted")
    return hits[0].parents[2]

WORK = Path("/kaggle/working/htd")
if WORK.exists():
    shutil.rmtree(WORK)
shutil.copytree(find_repo(), WORK)
os.chdir(WORK)
sys.path.insert(0, str(WORK / "src"))

MARKERS = {
    "scripts/train_aligntune.py": "--pair-types",
    "src/htd/train_data.py": "def filter_pair_types",
}
for rel, marker in MARKERS.items():
    assert marker in (WORK / rel).read_text(encoding="utf-8"), f"stale code dataset: {rel} lacks {marker!r}"
print("code markers ok")

def find_file(name):
    hits = [p for p in Path("/kaggle/input").rglob(name) if p.exists()]
    hits.sort(key=lambda p: len(p.parts))
    return hits[0] if hits else None

for p in Path("/kaggle/input").rglob("hf_token.txt"):
    os.environ.setdefault("HF_TOKEN", p.read_text().strip())
    os.environ.setdefault("HUGGINGFACE_API_KEY", p.read_text().strip())
assert os.environ.get("HF_TOKEN"), "attach rockerleo/htd-secrets"

DPO_DATA = find_file("dpo_train.jsonl")
print("DPO data:", DPO_DATA)
assert DPO_DATA, "attach rockerleo/htd-pairs-v2"
print(subprocess.run(["nvidia-smi", "--query-gpu=name,memory.total", "--format=csv"], capture_output=True, text=True).stdout)

def run(cmd, **kw):
    print("$", " ".join(str(c) for c in cmd), flush=True)
    return subprocess.run(cmd, check=True, **kw)

def assert_adapter(name):
    p = WORK / "outputs" / name
    cfg = p / "adapter_config.json"
    weights = p / "adapter_model.safetensors"
    assert cfg.exists(), f"{name}: adapter_config.json missing"
    assert weights.exists() and weights.stat().st_size > 1_000_000, f"{name}: adapter weights missing or empty"
    print(f"{name}: adapter ok")
```

```python
run([sys.executable, "-m", "pip", "install", "-q", "aligntune"])
```

```python
run(["aligntune", "info"], capture_output=True)
```

```python
run([sys.executable, "scripts/train_aligntune.py", "--data", str(DPO_DATA), "--pair-types", "drop,warn",
     "--output", "outputs/aligntune_drop_smoke", "--max-steps", "5", "--grad-accum", "2"])
assert_adapter("aligntune_drop_smoke")
# Check the training rows before the full run, not after it.
smoke_rows = [json.loads(x) for x in open(WORK / "outputs" / "aligntune_drop_smoke_train.jsonl") if x.strip()]
assert len(smoke_rows) == 858, f"expected 858 drop+warn rows, got {len(smoke_rows)}"
assert {r["pair_type"] for r in smoke_rows} == {"drop", "warn"}, "wrong pair types"
print("training rows ok: 858 drop+warn")
```

```python
# One changed axis from align-v2: lecture pairs removed (drop and warn kept).
run([sys.executable, "scripts/train_aligntune.py", "--data", str(DPO_DATA), "--pair-types", "drop,warn",
     "--output", "outputs/aligntune_dpo_drop", "--epochs", "1", "--lr", "5e-6", "--beta", "0.1", "--rank", "16"])
assert_adapter("aligntune_dpo_drop")
rows = sum(1 for x in open(WORK / "outputs" / "aligntune_dpo_drop_train.jsonl") if x.strip())
assert rows == 858, f"expected 858 drop+warn rows, got {rows}"
```
